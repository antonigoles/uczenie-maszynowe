```python
# Zadanie 1.2
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Wczytujemy sobie iris.data
# trzeba dodac naglowki bo w pliku nie ma
columns = ['sepal length','sepal width', 'petal length', 'petal width',  'class']
df = pd.read_csv("iris/iris.data", header=None, names=columns)

# Filtrujemy sobie tylko dla klasy setosa
df_setosa = df[df['class'] == 'Iris-setosa']

# Robimy sobie podział na przedziały
przedzialy = [0, 2.5, 3.0,  3.5, 4.0, 5.0]
etykiety = ['< 2.5', '2.5 - 3.0', '3.0 - 3.5', '3.5 - 4.0', '4.0+']
zliczenia = pd.cut(df_setosa['sepal width'], bins=przedzialy, right=False, labels=etykiety).value_counts().sort_index()

print("Liczba próbek w poszczególnych przedziałach:")
print(zliczenia)


# Wykres słupkowy
plt.figure(figsize=(8, 5))
zliczenia.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Rozkład sepal width dla Iris setosa')
plt.xlabel('Przedziały szerokości działki kielicha')
plt.ylabel('Liczba próbek')
plt.xticks(rotation=0)  # Ustawienie etykiet osi X w poziomie
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Zadanie 1.3
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Wczytujemy sobie iris.data
# trzeba dodac naglowki bo w pliku nie ma
columns = ['sepal length','sepal width', 'petal length', 'petal width',  'class']
df = pd.read_csv("iris/iris.data", header=None, names=columns)

# zescatterujmy najpierw kolejno każdy rodzaj
df_setosa = df[df['class'] == 'Iris-setosa']
df_versicolor = df[df['class'] == 'Iris-versicolor']
df_virginica = df[df['class'] == 'Iris-virginica']

skala_kropek = 15

plt.scatter(
    x=df_setosa['sepal length'], 
    y=df_setosa['sepal width'], 
    s=df_setosa['petal length'] * skala_kropek,
    c='red',
    label='setosa',
    alpha=0.6, # Dzieki temu nakładające sie będzie lepiej widać
)

plt.scatter(
    x=df_versicolor['sepal length'], 
    y=df_versicolor['sepal width'], 
    s=df_versicolor['petal length'] * skala_kropek,
    c='green',
    label='versicolor',
    alpha=0.6, # Dzieki temu nakładające sie będzie lepiej widać
)

plt.scatter(
    x=df_virginica['sepal length'], 
    y=df_virginica['sepal width'], 
    s=df_virginica['petal length'] * skala_kropek,
    c='blue',
    label='virginica',
    alpha=0.6, # Dzieki temu nakładające sie będzie lepiej widać
)

plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('(rozmiar = Petal Length)')
```

```python
# Zadanie 1.5

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Wczytujemy sobie iris.data
# trzeba dodac naglowki bo w pliku nie ma
columns = ['sepal length','sepal width', 'petal length', 'petal width',  'class']
df = pd.read_csv("iris/iris.data", header=None, names=columns)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten() # Spłaszczamy tablicę, żeby łatwo po niej iterować

# Lista naszych 4 atrybutów
atrybuty = ['sepal length', 'sepal width', 'petal length', 'petal width']

# Pętla rysująca wykres skrzypcowy dla każdego atrybutu osobno
for i, attr in enumerate(atrybuty):
    sns.violinplot(
        x='class',         # Oś X: podział na gatunki irysa
        y=attr,            # Oś Y: wartości danego atrybutu
        hue='class',
        data=df,           # Nasz DataFrame
        ax=axes[i],        # Przypisanie do konkretnego okienka w siatce
        palette='Set2',    # Paleta kolorów
        legend=False,      # Wyłączenie zbędnej legendy (skoro oś X już to opisuje)
        inner='box'        # Pokazanie środkowego pudełka (box plot) wewnątrz skrzypiec
    )
    
    axes[i].set_title(f'Wykres skrzypcowy: {attr}')
    axes[i].set_xlabel('Gatunek')
    axes[i].set_ylabel(attr)
    axes[i].grid(True, linestyle='--', alpha=0.5)

# Dopasowanie odstępów i wyświetlenie
plt.tight_layout()
plt.show()
```

```python
# Zadanie 2.1 (dla próbek pozytywnych i negatywnych)
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("bank_marketing/bank.csv", sep=";")
df_yes = df[df['y'] == 'yes']
df_no = df[df['y'] == 'no']

# Bedziemy mieli dwa wykresy obok siebie
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# 1. Próbki negatywne (no)
axes[0].hist(df_no['duration'], bins=50, color='salmon', edgecolor='black')
axes[0].set_title('Histogram duration: y = "no" (Negatywne)')
axes[0].set_xlabel('Duration (s)')
axes[0].set_ylabel('Liczba próbek')
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Próbki pozytywne (yes)
axes[1].hist(df_yes['duration'], bins=50, color='lightgreen', edgecolor='black')
axes[1].set_title('Histogram duration: y = "yes" (Pozytywne)')
axes[1].set_xlabel('Duration (s)')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Zadanie 2.2
import pandas as pd
import matplotlib.pyplot as plt

import ipywidgets as widgets
from ipywidgets import interact

# Do tego zadania zaciągne full liste
df = pd.read_csv("bank_marketing/bank-full.csv", sep=";")

@interact(min_age=widgets.IntSlider(min=18, max=80, step=1, value=25, description='Min wiek:'))
def plot_balance_by_age(min_age):
    df_subset = df[df['age'] > min_age]
    plt.figure(figsize=(10, 5))
    plt.hist(df_subset['balance'], bins=50, color='skyblue', edgecolor='black')
    plt.title(f'Histogram atrybutu balance dla osób o wieku > {min_age} (Liczba osób: {len(df_subset)})')
    plt.xlabel('Balance (saldo)')
    plt.ylabel('Liczba próbek')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
    
```

```python
# Zadanie 2.3
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("bank_marketing/bank.csv", sep=";")

# Progi co 20 sekund
progi_ls = list(range(0, 5500, 20))
progi = []
procenty_pozytywnych = []
for t in progi_ls:
    subset = df[df['duration'] > t]
    if len(subset) > 0:
        pozytywne = len(subset[subset['y']  == 'yes'])
        procent = (pozytywne / len(subset)) * 100
    else:
        break
    progi.append(t)
    procenty_pozytywnych.append(procent)

plt.figure(figsize=(10, 5))
plt.plot(progi, procenty_pozytywnych, marker='o', linestyle='-', color='crimson', linewidth=2, markersize=5)

plt.title('Zależność procentu próbek pozytywnych od minimalnego czasu trwania rozmowy (t)')
plt.xlabel('Próg t (minimalny czas duration w sekundach)')
plt.ylabel('Procent próbek pozytywnych (%)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Zadanie 2.4
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("bank_marketing/bank-full.csv", sep=";")

# Progi co 100 EUR
progi_ls = list(range(0, 106500, 100))
progi = []
procenty_pozytywnych = []
for t in progi_ls:
    subset = df[df['balance'] > t]
    if len(subset) > 0:
        pozytywne = len(subset[subset['y']  == 'yes'])
        procent = (pozytywne / len(subset)) * 100
    else:
        break
    progi.append(t)
    procenty_pozytywnych.append(procent)

plt.figure(figsize=(10, 5))
plt.plot(progi, procenty_pozytywnych, marker='o', linestyle='-', color='crimson', linewidth=2, markersize=5)

plt.title('Zależność procentu próbek pozytywnych od minimalnego średniego rocznego balansu na koncie')
plt.xlabel('Próg balansu na koncie (minimalny srednia roczna ilosc piniędzy na koncie w euro)')
plt.ylabel('Procent próbek pozytywnych (%)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Zadanie 2.5
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("bank_marketing/bank-full.csv", sep=";")
# Podzial po równo
df_train = df.sample(frac=0.5, random_state=17)
df_test = df.drop(df_train.index)

def blad_w_zaleznosci_od_progu(prog, df):
    mask_above = df['balance'] > prog
    mask_below = df['balance'] <= prog
    
    count_above_threshold = mask_above.sum()
    count_above_threshold_negative = ((mask_above) & (df['y'] == 'no')).sum()
    
    count_below_equal = mask_below.sum()
    count_below_equal_positive = ((mask_below) & (df['y'] == 'yes')).sum()
    
    above_percent_success = (1 - (count_above_threshold_negative / count_above_threshold)) if count_above_threshold > 0 else 0
    below_equal_percent_success = (count_below_equal_positive / count_below_equal) if count_below_equal > 0 else 0

    # Blad powinien rosnac wraz z below_equal_percent_success i malec wraz z above_percent_success czyli
    return (1.0 - above_percent_success) + below_equal_percent_success

wyniki_train = []

progi = list(range(0, 10001, 200))
for p in progi:
    res_train = blad_w_zaleznosci_od_progu(p, df_train)
    wyniki_train.append(res_train)

plt.figure(figsize=(10, 5))
plt.plot(progi, wyniki_train, label='Zbiór treningowy', color='blue', linestyle='-')

plt.title('Zależność miary błędu/skuteczności od progu salda (balance)')
plt.xlabel('Próg balance')
plt.ylabel('Wartość metryki')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


print("Blad do danych testowych zakladajac ze wezmiemy prog ktory jest minimum globalnym wykresu: ", blad_w_zaleznosci_od_progu(1700, df_test))
```
